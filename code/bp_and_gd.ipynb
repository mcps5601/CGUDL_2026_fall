### 權重命名對照

投影片使用 $w_{i,j}$ 的雙下標，notebook 為求簡潔寫成 `wij`，兩者一一對應：

| 投影片 | Notebook | 連線 | 出現在 |
|:---:|:---:|:---:|---|
| $w_{1,1}$ | `w11` | $x_1 \rightarrow h_1$ | $h_1 = w_{1,1}x_1 + w_{2,1}x_2 + b_1$ |
| $w_{2,1}$ | `w21` | $x_2 \rightarrow h_1$ | 同上 |
| $w_{1,2}$ | `w12` | $x_1 \rightarrow h_2$ | $h_2 = w_{1,2}x_1 + w_{2,2}x_2 + b_2$ |
| $w_{2,2}$ | `w22` | $x_2 \rightarrow h_2$ | 同上 |
| $w_{3,1}$ | `w31` | $h_1 \rightarrow \hat{y}$ | $\hat{y} = w_{3,1}h_1 + w_{3,2}h_2 + b_3$ |
| $w_{3,2}$ | `w32` | $h_2 \rightarrow \hat{y}$ | 同上 |

- 第 1 層的下標讀作 **(來源輸入, 目標隱藏節點)**：$w_{2,1}$ 是「$x_2$ 連到 $h_1$」。
- 輸出層的第一個下標 3 代表**第 3 層（輸出層）**，第二個下標才是來源：$w_{3,2}$ 是「$h_2$ 連到 $\hat{y}$」。
- 程式中的梯度變數同理，例如 `dL_dw21` 即 $\dfrac{\partial \mathcal{L}}{\partial w_{2,1}}$。

### 定義數值

```python
# 1. Input and ground truth
x1 = 1.0
x2 = 2.0
y = 3.0

# Weights
w11 = 0.5
w12 = 0.4
w21 = 0.3
w22 = 0.2
w31 = 0.6
w32 = 0.7

# Bias
b1 = 0.1
b2 = 0.1
b3 = 0.1
```

### 正向傳播

```python
# 2. Forward pass

h1 = w11 * x1 + w21 * x2 + b1
h2 = w12 * x1 + w22 * x2 + b2

y_hat = w31 * h1 + w32 * h2 + b3

loss = 0.5 * (y - y_hat) ** 2

print("h1 =", h1)
print("h2 =", h2)
print("y_hat =", y_hat)
print("loss =", loss)
```

### 反向傳播

![反向傳播法觀察 (1)，對應 slide p.32](https://raw.githubusercontent.com/mcps5601/CGUDL_2026_fall/main/assets/w3_p32.png)

```python
# 3. Backward pass (可對照 slide: p29)

# dL / dy_hat
dL_dyhat = -(y - y_hat)

# Layer 2
dL_dw31 = h1 * dL_dyhat
dL_dw32 = h2 * dL_dyhat

print("dL/dw31 =", dL_dw31)
print("dL/dw32 =", dL_dw32)
```

![Backpropagation (Layer: 1, w_1,1)，對應 slide p.36](https://raw.githubusercontent.com/mcps5601/CGUDL_2026_fall/main/assets/w3_p36.png)

```python
# 4-1. Gradient passed backward to h1 and h2

dL_dh1 = w31 * dL_dyhat
dL_dh2 = w32 * dL_dyhat
```

```python
# 4-2. Layer 1
dL_dw11 = x1 * dL_dh1
dL_dw21 = x2 * dL_dh1

dL_dw12 = x1 * dL_dh2
dL_dw22 = x2 * dL_dh2
```

### 梯度下降

```python
# 5. 進行梯度下降以更新權重（省略 bias）

lr = 0.1

w11 = w11 - lr * dL_dw11
w12 = w12 - lr * dL_dw12
w21 = w21 - lr * dL_dw21
w22 = w22 - lr * dL_dw22

w31 = w31 - lr * dL_dw31
w32 = w32 - lr * dL_dw32
```

### 更新後結果

```python
# 6. 再做一次 forward pass

h1 = w11 * x1 + w21 * x2 + b1
h2 = w12 * x1 + w22 * x2 + b2
y_hat = w31 * h1 + w32 * h2 + b3

new_loss = 0.5 * (y - y_hat) ** 2

print("Old loss:", loss)
print("New loss:", new_loss)
```
